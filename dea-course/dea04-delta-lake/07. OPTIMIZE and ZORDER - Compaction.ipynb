## Compaction using OPTIMIZE and Z-Ordering


### OPTIMIZE Command
Merges multiple small files into fewer, larger files, thus improves the performance

It doesn't remove any history, all the history is still being kept.

#### 1. Create table - demo.delta_lake.optimized_stock_prices

```sql
DROP TABLE IF EXISTS demo.delta_lake.optimized_stock_prices;

CREATE TABLE IF NOT EXISTS demo.delta_lake.optimized_stock_prices
(
  stock_id STRING,
  price DOUBLE,
  trading_date DATE
);
```

#### 2. Inserts some data in a few transactions

```sql
INSERT INTO demo.delta_lake.optimized_stock_prices
VALUES ('AAPL', 227.65, "2025-02-18");
```

```sql
INSERT INTO demo.delta_lake.optimized_stock_prices
VALUES ('GOOGL', 2275.0, "2025-02-18");
```

```sql
INSERT INTO demo.delta_lake.optimized_stock_prices
VALUES ('MSFT', 325.0, "2025-02-10");
```

```sql
INSERT INTO demo.delta_lake.optimized_stock_prices
VALUES ('AMZN', 3520.0, "2025-02-12");
```

```sql
SELECT * FROM demo.delta_lake.optimized_stock_prices;
```

#### 3. Check the table history

```sql
DESC HISTORY demo.delta_lake.optimized_stock_prices;
```

#### 4. Check the number of files required to get the latest data

```sql
DESC DETAIL demo.delta_lake.optimized_stock_prices;
```

```sql
SELECT * FROM demo.delta_lake.optimized_stock_prices;
```

#### 5. Run OPTIMIZE

```sql
OPTIMIZE demo.delta_lake.optimized_stock_prices;
```

```sql
DESC HISTORY demo.delta_lake.optimized_stock_prices;
```

```sql
SELECT * FROM demo.delta_lake.optimized_stock_prices VERSION AS OF 3;
```

### ZORDER

```sql
OPTIMIZE demo.delta_lake.optimized_stock_prices
ZORDER BY stock_id;
```
