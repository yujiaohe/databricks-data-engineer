## Remove usused files using VACCUM


### VACCUM Command
Used to remove old, unused files from the Delta Lake to free up storage. Permanently deletes files that no longer referenced in the transaction log and older than retention threshold (default 7 days)

History will be deleted.

#### 1. Check the table history

```sql
DESC HISTORY demo.delta_lake.optimized_stock_prices;
```

#### 2. Run VACCUM

```sql
VACUUM demo.delta_lake.optimized_stock_prices;
```

```sql
DESCRIBE HISTORY demo.delta_lake.optimized_stock_prices;
```

No files are deleted, since default retention is 7 days.

```sql
-- Don't do it in production
SET spark.databricks.delta.retentionDurationCheck.enabled = false;
VACUUM demo.delta_lake.optimized_stock_prices RETAIN 0 HOURS;
```

```sql
SELECT * FROM demo.delta_lake.optimized_stock_prices;
```

```sql
DESC HISTORY demo.delta_lake.optimized_stock_prices;
```

#### VACUUM command common scenarios:
- 7 days intervals 
- 28 days to keep hold of customer data.
