## Extract Data From the Address Files

1. Demonstrate Limitations of CSV file format in SELECT statement
2. User read_files function to overcome the limitations
3. Create Addresses view in the Bronze Layer


### 1. Demonstrate Limitations of CSV file format in SELECT statement

```sql
SELECT * FROM CSV.`/Volumes/gizmobox/landing/operational_data/addresses/`
```


### 2. User read_files function to overcome the limitations

[read_files()](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/read_files)

```sql
SELECT 
    *,
    _metadata.file_name
  FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses', 
                  format =>'csv', 
                  header => True, 
                  delimiter => '\t');
```


### 3. Create Addresses view in the Bronze Layer

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_addresses 
AS
SELECT 
    *,
    _metadata.file_name
  FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses', 
                  format =>'csv', 
                  header => True, 
                  delimiter => '\t');
```

```sql
SELECT * FROM gizmobox.bronze.v_addresses;
```
