## Extract Data From the Payments Files

1. List the files from Payment folder
2. Create External Table
3. Demonstrate the effect of Adding/Updating/Deleting files
4. Demonstrate the effect of Dropping the Table

### 1. List the files from Payment folder

```sql
%fs ls 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing/external_data/payments'
```

### 2. [Create External Table](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/sql-ref-syntax-ddl-create-table-using)

- `LOCATION` makes the table external

```sql
CREATE TABLE IF NOT EXISTS gizmobox.bronze.payments
  (payment_id INTEGER, order_id INTEGER, payment_timestamp TIMESTAMP, payment_status INTEGER, payment_method STRING)
USING CSV
OPTIONS (
  header = "true",
  delimiter = ","
)
LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing/external_data/payments';
```

```sql
SELECT * FROM gizmobox.bronze.payments;
```

```sql
DESCRIBE EXTENDED gizmobox.bronze.payments;
```

### 3. Demonstrate the effect of Adding/Updating/Deleting files

```sql
SELECT * FROM gizmobox.bronze.payments;
```

```sql
REFRESH TABLE gizmobox.bronze.payments;
```

```sql
SELECT * FROM gizmobox.bronze.payments;
```

```sql
%fs ls 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing/external_data/payments'
```

### 4. Demonstrate the effect of Dropping the Table

```sql
DROP TABLE IF EXISTS gizmobox.bronze.payments;
```

## Important Note:

Whenever you are using external tables to read the data, make sure to refresh the table so that it's picking up all the new data that's available.
