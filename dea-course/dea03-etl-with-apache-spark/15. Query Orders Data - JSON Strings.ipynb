## Query Orders Data as JSON Strings
1. Extract Top Level Column Values
2. Extract Array elements
3. Extract Nested Column Values
4. CAST Column Values to a Specific Data Type

```sql
SELECT *
FROM gizmobox.bronze.v_orders;
```

### 1. Extract Top Level Object Values
Top Level Object `<column_name>:<extraction_path>`

```sql
SELECT 
  value:order_id AS order_id,
  value
FROM gizmobox.bronze.v_orders;
```

### 2. Extract Array elements
To extract array elements - use the indices in the brackets `<column_name>:<extraction_path>[index]`

```sql
SELECT 
  value:order_id AS order_id,
  value:items[0] as item_1,
  value:items[1] as item_2,
  value
FROM gizmobox.bronze.v_orders;
```

### 3. Extract Nested Column Values
To extract nested fields - use tdot notation `<column_name>:<extraction_path>[index].<leaf node>`

```sql
SELECT 
  value:items[0].item_id as item_1_id,
  value:items[0]as item_1,
  value
FROM gizmobox.bronze.v_orders;
```

### 4. CAST Column Values to a Specific Data Type

```sql
SELECT 
  value:items[0].item_id::INT item_1_id,
  value:items[0] AS item_1,
  value
FROM gizmobox.bronze.v_orders;
```
