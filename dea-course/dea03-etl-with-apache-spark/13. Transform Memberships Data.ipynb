## Transform Memberships Data
1. Extract customer_id from the file path
2. Write transformed data to the Silver schema

```sql
SELECT * FROM gizmobox.bronze.v_memberships;
```

### 1. Extract customer_id from the file path

[Documentation of regexp_extract](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/regexp_extract)

[Regexp Pattern](https://docs.oracle.com/javase/8/docs/api/java/util/regex/Pattern.html)


```sql
SELECT
  regexp_extract(path, '.*/(\\d+)\\.png$', 1) AS customer_id,
  content AS membership_card
FROM gizmobox.bronze.v_memberships;
```

### 2. Write transformed data to the Silver schema

```sql
CREATE TABLE IF NOT EXISTS gizmobox.silver.memberships
AS
SELECT
  regexp_extract(path, '.*/(\\d+)\\.png$', 1) AS customer_id,
  content AS membership_card
FROM gizmobox.bronze.v_memberships;
```

```sql
SELECT * FROM gizmobox.silver.memberships;
```

```sql
DESCRIBE EXTENDED gizmobox.silver.memberships;
```
