## Set-up the project environment for GizmoBox Data Lakehouse

### 1. Access the container gizmobox

```sql
%fs ls "abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing/operational_data/"
```

### 2. Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_gizmobox
    URL "abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/"
    WITH (STORAGE CREDENTIAL dea_course_ext_sc)
    COMMENT "External Location for Gizmobox Data Lakehouse"
```

### 3. Create the catalog - gizmobox

[Azure Databricks Create Catalog](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/sql-ref-syntax-ddl-create-catalog)

```sql
SHOW CATALOGS
```

```sql
CREATE CATALOG IF NOT EXISTS gizmobox
      MANAGED LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/'
      COMMENT "This is the catalog for the Gizmobox Data Lakehouse";
```

### 4. Create Schemas

- Landing
- Bronze
- Silver
- Gold

```sql
SELECT current_catalog()
```

```sql
USE CATALOG gizmobox;


CREATE SCHEMA IF NOT EXISTS landing
      MANAGED LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing'
      COMMENT "This is the landing schema for the Gizmobox Data Lakehouse";

CREATE SCHEMA IF NOT EXISTS bronze
      MANAGED LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/bronze'
      COMMENT "This is the bronze schema for the Gizmobox Data Lakehouse";

CREATE SCHEMA IF NOT EXISTS silver
      MANAGED LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/silver'
      COMMENT "This is the silver schema for the Gizmobox Data Lakehouse";

CREATE SCHEMA IF NOT EXISTS gold
      MANAGED LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/gold'
      COMMENT "This is the gold schema for the Gizmobox Data Lakehouse";
```

```sql
SHOW SCHEMAS;
```

#### 5. Create Volume

```sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://gizmobox@deacourseextsa111.dfs.core.windows.net/landing/operational_data/'
```
