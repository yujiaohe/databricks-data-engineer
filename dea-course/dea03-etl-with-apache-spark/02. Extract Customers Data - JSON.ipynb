
## Extract Data From Customers JSON File

1. Query Single File
2. Query List of FIles using wildcard Characters
3. Query all the files in a Folder
4. Select File Metadata
5. Register Files in Unity Catalog using Views
6. Create Temporary View
7. Create Global Temporary View



### 1. Query Single JSON File

```sql
%fs ls '/Volumes/gizmobox/landing/operational_data/customers'
```

```sql
SELECT * FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json`
```


### 2. Query Multiple JSON Files

```sql
SELECT * FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024_*.json`
```


### 3. Query all the files in a folder

```sql
SELECT * FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```


### 4. Select File Metadata

```sql
SELECT input_file_name() as file_path, -- Deprecated from Databricks Runtime 13.3 LTS onwards
        * 
  FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
SELECT _metadata.file_path,
        * 
  FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```

### 5. Register Files in Unity Catalog using Views

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers
AS
SELECT *,
       _metadata.file_path
  FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
SELECT * FROM gizmobox.bronze.v_customers
```

### 6. Create Temporary View

- Concept within spark
- Exists within spark session
- Destroyed as soon as the spark session ends, that is when the cluster is detached

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_customers
AS
SELECT *,
       _metadata.file_path
  FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
SELECT * FROM tv_customers;
```

### 7. Create Global Temporary View

- Exists within a temporary system schema called `global_temp`
- Destroyed when the spark application ends, that is when the cluster is terminated

```sql
CREATE OR REPLACE GLOBAL TEMPORARY VIEW gtv_customers
AS
SELECT *,
       _metadata.file_path
  FROM JSON.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
SELECT * FROM global_temp.gtv_customers;
```
