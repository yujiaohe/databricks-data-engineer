## Create Customer Order Summary
1. Join the tables Silver_Customers, Silver_Addresses and Silver_Orders
2. Retrieve the latest Address of the Customer
3. Calculate the following values
   - total_orders
   - total_items_ordered
   - total_order_amount

Customer Order Summary Requirement
![](./images/customer-order-summary.png)

```sql
CREATE OR REFRESH MATERIALIZED VIEW gold_customer_order_summary
AS
SELECT
  c.customer_id,
  c.customer_name,
  c.date_of_birth,
  c.telephone,
  c.email,
  a.address_line_1,
  a.city,
  a.state,
  a.postcode,
  COUNT(DISTINCT o.order_id) AS total_orders,
  SUM(o.item_quantity) AS total_items_ordered,
  SUM(o.item_price * o.item_quantity) AS total_order_amount
FROM LIVE.silver_customers AS c
  JOIN LIVE.silver_addresses AS a ON a.customer_id = c.customer_id
  JOIN LIVE.silver_orders AS o ON o.customer_id = c.customer_id
WHERE
  a.__END_AT IS NULL
GROUP BY ALL; --shortcut of all columns except calculated
```
