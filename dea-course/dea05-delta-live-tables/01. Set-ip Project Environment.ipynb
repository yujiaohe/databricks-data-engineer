## Set-ip the project envrionment for CircuitBox Data Lakehouse
1. Create external location - dea_course_ext_dl_circuitbox
2. Create Catolog - circuitbox
3. Create Schemas 
    - landing
    - lakehouse
4. Create Volume - operational_data

![data-architecture](./images/data-architecture.png)


## 1. Create External Location

**External Location Name**: `dea_course_ext_dl_circuitbox`</br>
ADLS Path: `abfss://circuitbox@deacourseextsa111.dfs.core.windows.net` </br>
Storage Credential: `dea_course_ext_sc`

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
URL 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL dea_course_ext_sc)
COMMENT 'External location for the circuitbox data lakehouse';
```

```sql
%fs ls 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net'
```

## 2. Create Catalog
**Catalog Name**: `circuitbox` </br>
Managed Location: `abfss://circuitbox@deacourseextsa111.dfs.core.windows.net`

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
MANAGED LOCATION 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net'
COMMENT 'Catalog for the circuitbox data lakehouse';
```

```sql
SHOW CATALOGS;
```

## 3. Create Schemas
  1. **Schema Name**: `landing` </br>
      Managed Location: `abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/landing/`
  2. **Schema Name**: `lakehouse` </br>
      Managed Location: `abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/lakehouse/`

```sql
USE CATALOG circuitbox;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS lakehouse
MANAGED LOCATION 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/lakehouse';
```

```sql
SHOW SCHEMAS;
```

## 4. Create Volume
**Volume Name**: operational_data </br>
ADLS Path: `abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/landing/operational_data`

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS circuitbox.landing.operational_data
LOCATION 'abfss://circuitbox@deacourseextsa111.dfs.core.windows.net/landing/operational_data';
```

```sql
%fs ls '/Volumes/circuitbox/landing/operational_data'
```
